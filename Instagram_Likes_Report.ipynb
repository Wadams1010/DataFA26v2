I unfortunately do not have an account on either Instagram or Facebook, so the source of my data comes from my teacher's Instagram account. <br>
This data was created by Instagram in order to keep track of the likes applied by this account in order to personalize their experience in the app. <br>

This data may be unreliable because there exists certain aspects of the data that are not provided here. <br>

```python
import json
import pandas as pd

with open("liked_posts.json", "r") as f:
    data = json.load(f)

data[0]
```

```python
row = {}
row['timestamp'] = data[0]['timestamp']

row
```

```python
for item in data[0]['label_values']:
    if 'label' in item:
        row[item['label']] = item['value']

row
```

```python
for item in data[0]['label_values']:
    if 'label' in item:
        row[item['label']] = item['value']
    elif item.get('title') == 'Owner':
        owner_fields = item['dict'][0]['dict']
        for owner_item in owner_fields:
            row['Owner ' + owner_item['label']] = owner_item['value']

row
```

```python
tidy_rows = []

for entry in data:
    row = {}
    row['timestamp'] = entry['timestamp']
    
    for item in entry['label_values']:
        if 'label' in item:
            row[item['label']] = item['value']
        elif item.get('title') == 'Owner':
            owner_fields = item['dict'][0]['dict']
            for owner_item in owner_fields:
                row['Owner ' + owner_item['label']] = owner_item['value']
    tidy_rows.append(row)

tidy_rows[:2]
```

My teacher has mentioned several hobbies of hers, the two most prominant being fight robots and golfing. I would like to see if there is evidence of these activities in her Instagram data, as well as if there is evidence of her job as a teacher at Kent State.

I believe that the columns that I have chosen will allow me to answer this question, as the caption and owner name/username will include data that will allow me to see what the post, or account that made said post, is about. I will be sorting the data by date and taking the ten most recent posts, as the posts that are most recent are also likely to provide the most accurate view of the user and their account.

```python
likes_df = pd.DataFrame(tidy_rows)
likes_df.head(10)
```

```python
likes_df['DateTime'] = pd.to_datetime(likes_df['timestamp'], unit='s')
likes_df[['timestamp', 'DateTime']].head()
```

One drawback of using this kind of data is seen here; the most recent posts within it are already outdated by several weeks at time of writing. In order to keep it up-to-date, one would need to extract this data from Instagram constantly, which is a very inefficient and time-consuming process.
