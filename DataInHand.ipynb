# Data In Hand - Steam Games

The data I will be using comes from this website : https://www.kaggle.com/datasets/fronkongames/steam-games-dataset/data

This is a .json dataset containing various information about over 125,000 games that have been published on Steam. It is organized by the unique app id number of each game, which contains information such as it's name, release date, price, estimated number of owners, language, and genre, among many other fields that will allow me to test my hypothesis.

The data itself was collected by a dedicated scraper that uses an API provided by Steam. The dataset is updated weekly and has already been used to create accurate analysis reports regarding the trends and insights of Steam games. Considering that I plan to create an analysis report of my own that will discuss very similar subjects, I know that this dataset will provide me the information that I need.
